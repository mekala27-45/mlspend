# A low threshold exceeded the false-alarm budget

A seasonal anomaly detector still needs calibration. Inspect the already-recorded development search: the lowest threshold looked responsive but exceeded the registered non-incident alarm budget in real simulated series.

This is a reproducible analysis of an actual failed approach in this build. It is not a backdated research log.

```python
from pathlib import Path
import json
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p/'packages').exists())
sys.path.insert(0,str(ROOT))
m=json.loads((ROOT/'results/expanded/manifest.json').read_text())
plan=json.loads((ROOT/'evaluation/v2/plan.json').read_text())
budget=plan['detector']['false_alarm_budget']
failures=[{'team':s['team'],'service':s['service'],'grain':s['grain'],**v} for s in m['detection']['series'] for v in s.get('curve',[]) if v['threshold']==min(plan['detector']['threshold_candidates']) and v['false_alarm_rate']>budget]
print(json.dumps(failures[:8],indent=2))
print({'failing_development_candidates':len(failures),'budget':budget})
assert failures
```

Dead end: selecting the smallest threshold because it catches a visible spike. Keep the predeclared false-alarm budget, expose series whose chosen point is on the boundary, and do not tune against held-out incidents.
