# A cheaper spot rate increased the platform bill

The first quiet development hours provide a counterexample to pricing a move from the spot rate alone. Rerun the same demand with and without the spot action; minimum shared capacity can retain the on-demand bill.

This is a reproducible analysis of an actual failed approach in this build. It is not a backdated research log.

```python
from pathlib import Path
import json
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p/'packages').exists())
sys.path.insert(0,str(ROOT))
m=json.loads((ROOT/'results/expanded/manifest.json').read_text())
from packages.prices.book import PriceBook
from packages.simulator.expanded import simulate
book=PriceBook(ROOT/'data/prices/expanded')
baseline=sum(r['cost_micro'] for r in simulate(book,hours=3))
spot=sum(r['cost_micro'] for r in simulate(book,hours=3,actions=frozenset({'spot'})))
print({'baseline_micro':baseline,'spot_micro':spot,'whole_platform_saving_micro':baseline-spot})
assert spot > baseline
```

Dead end: multiplying the released allocation by a cheaper GPU rate and calling the difference cash. The full study therefore reruns every action and reports capacity externalities and joint effects separately.
